```python
!pip install -q transformers torch

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# 1. Load the tokenizer and model explicitly (This bypasses the pipeline error)
model_name = "google/flan-t5-base"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# 2. Raw financial earnings text (unstructured data)
financial_text = """
AlphaCorp Financial reported total revenue of €45 million, representing a 15% increase year-over-year.
Growth was primarily driven by strong performance in retail banking and wealth management.
Operating expenses rose by 8% due to investments in cloud infrastructure and cybersecurity.
The executive team confirmed that strategic priorities for next quarter focus on operational efficiency
and launching an AI-driven digital assistant for customer accounts.
"""

# 3. Instruction prompt for extracting key metrics
prompt = f"""
Read the text and answer each point:
1. Total revenue:
2. Revenue growth rate:
3. Key revenue drivers:
4. Reason for higher expenses:
5. Strategic priority:

Text: {financial_text}
"""

# 4. Tokenize the input and run model inference
inputs = tokenizer(prompt, return_tensors="pt")
outputs = model.generate(**inputs, max_new_tokens=150)

# 5. Decode and display the result
result = tokenizer.decode(outputs[0], skip_special_tokens=True)
print("--- AI EXTRACTION OUTPUT ---")
print(result)
```
